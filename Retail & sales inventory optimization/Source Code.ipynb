```python
import pandas as pd
import numpy as np

# 1. Load the raw dataset
df = pd.read_csv('Retail_sales.csv.csv')
print(f"Initial Dataset Shape -> Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# 2. Data Cleaning & Preprocessing
# Remove duplicate records
df = df.drop_duplicates()

# Handle missing values
df = df.dropna()

# 3. Save the Cleaned Dataset (Columns will remain exactly 31)
cleaned_filename = 'Cleaned_Retail_Sales.csv'
df.to_csv(cleaned_filename, index=False)

print("\nData cleaning complete!")
print(f"Cleaned file saved successfully as: {cleaned_filename}")
print(f"Final Cleaned Data Shape -> Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```
